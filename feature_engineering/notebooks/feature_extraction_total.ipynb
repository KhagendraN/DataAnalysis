```python
import pandas as pd

class FeatureExtractionTotal:
    """Extracts frame for total budget and total expenditure"""
    
    def __init__(self, budget_frame: pd.DataFrame, exp_frame: pd.DataFrame):
        self.budget_frame = budget_frame
        self.exp_frame = exp_frame

    def extract_index_for_total(self) -> list:
        """Extract index for total and return list"""
        index_names = [
            "Province 1 Total",
            "Madesh Province Total",
            "Bagmati Province Total",
            "Gandaki Province Total",
            "Lumbini Province Total",
            "Karnali Province Total",
            "Sudurpaschim Province Total"
        ]
        indexes = []
        
        for idx, province in enumerate(index_names):
            # Get rows matching the current province
            province_rows = self.budget_frame[self.budget_frame["SN"] == province]
            
            if province_rows.empty:
                # If province is missing, print a message and skip this iteration
                print(f"Name is missing: {province}")
                continue  # Skip this province if it's missing
            
            # Get the index of the first matching row
            index_value = province_rows.index[0]
            indexes.append(index_value)
        
        return indexes
    
    def extract_total(self):
        # Get the indexes for the total provinces
        indexes = self.extract_index_for_total()
        
        print("Index extracted 😍")
        
        # Extract total budget and expenditure frames based on the indexes
        total_budget_df = self.budget_frame.iloc[indexes]
        total_exp_df = self.exp_frame.iloc[indexes]
        
        print("Frame for total extracted 😍")
        
        # Rename the 'SN' column to 'Province' for clarity
        total_budget_df = total_budget_df.rename(columns={"SN": "Province"})
        total_exp_df = total_exp_df.rename(columns={"SN": "Province"})
        
        # Remove the 'Local Body' column as it is not required
        total_budget_df = total_budget_df.drop("Local Body", axis=1, errors='ignore')
        total_exp_df = total_exp_df.drop("Local Body", axis=1, errors='ignore')
        
        print("Successful 😍")
        
        return total_budget_df, total_exp_df
```
