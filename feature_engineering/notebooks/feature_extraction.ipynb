### Feature extarction for each province
- Here we'll observe the data structure and extract important features 
- Budget and expenditure will be separated 

| Attribute Name | Category        | Meaning and Context                                                                                                                                                      |
|----------------|-----------------|--------------------------------------------------------------------------------------------------------------------------------------------------------------------------|
| SN             | Identifier      | Serial Number or Sl. No. A simple numerical index to count the entries (local bodies/rows).                                                                              |
| Local Body     | Identifier      | The name of the specific local government unit (e.g., Municipality, Rural Municipality, District Coordination Committee) whose financial data is being reported.           |
| Recurrent      | Budget/Expenditure | Spending on day-to-day operations and administrative costs that recur every year. This includes salaries, office supplies, utilities, maintenance, and regular subsidies. |
| Capital        | Budget/Expenditure | Spending on creating or acquiring long-term assets. This is investment spending, such as building roads, schools, hospitals, water systems, or purchasing land and heavy machinery. |
| Financial      | Budget/Expenditure | Spending related to financial transactions. This often includes interest payments on loans, principal loan repayments, and other financial liabilities.                    |
| Total          | Summary         | The sum of the respective Recurrent, Capital, and Financial categories. Since you have two sets of these (Recurrent, Capital, Financial, Total), this column represents the Total Budget or Total Expenditure. |
| Cash Balance   | Balance         | The closing balance of cash or bank reserves at the end of the reporting period (e.g., end of the fiscal year) after all expenditures have been accounted for.               |
