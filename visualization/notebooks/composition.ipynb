```python
# Attribute names
attributes = ['a1', 'a2', 'a3', 'a4', 'a5', 'a6', 'a7']

# Sample data for each attribute (total amount and expenses for each attribute)
# Each entry represents different expense categories (e.g., expense1, expense2, expense3)
# For simplicity, let's assume 3 expense categories.
expense1 = [10, 15, 20, 25, 30, 35, 40]
expense2 = [5, 10, 15, 10, 20, 25, 30]
expense3 = [3, 7, 8, 5, 10, 15, 20]

# Stack expenses to get the total amount for each bar
total_amount = np.array(expense1) + np.array(expense2) + np.array(expense3)

# Plot
fig, ax = plt.subplots(figsize=(8, 4))

# Stacked bars
ax.barh(attributes, expense1, color='lightblue', label='Expense 1')
ax.barh(attributes, expense2, left=expense1, color='lightgreen', label='Expense 2')
ax.barh(attributes, expense3, left=np.array(expense1) + np.array(expense2), color='salmon', label='Expense 3')

# Add total labels
for i, (t, e1, e2, e3) in enumerate(zip(total_amount, expense1, expense2, expense3)):
    ax.text(t + 2, i, str(t), va='center', ha='left', fontsize=10)

# Labels and title
ax.set_xlabel('Total Amount')
ax.set_ylabel('Attributes')
ax.set_title('Stacked Bar Chart with Expenses')

# Add legend
ax.legend()

# Show plot
plt.show()
```
